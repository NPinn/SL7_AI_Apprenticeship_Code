## Deep learning vs traditional ML
* Traditional ML: handcrafted features + ML (hand-engineered features are time consuming, brittle and not scalable in practice)
* Deep Learning: learning the relevant features directly (representation learning)


Deep learning has one really important benefit compared to traditional ML: it is able to automatically learn the relevant features (representation learning), so we do not need to hand-engineer them. Engineering the features is sometimes very complicated, for example in computer vision.

## What to achieve from data? Generalization!

* Deep learning models have the ability to over-fit on their training data. it is useful to think of this as the model memorizing the training data
* The model might then perform well on the training data - but badly on new data (for example a partitioned test data set)
* This is a failure of the model the generalize well
* We monitor this over-fitting by partitioning data into three parts:
  * training - used to optimize model weights
  * validation - used to test for over-fitting during training
  * test - used to measure the final performance of the model

We will cover the details of how a NN is trained in the next session. For now, let's focus on the higher-level goal of what we want to achieve. We want to achieve excellent generalization ability. We use the available data to do two main things:

Training Data - Learn a weight vector (neural network model) that **maximizes** the fit to data (or rather, **minimizes** the loss)
Testing Data - Measure the performance for the task it has been trained for

Note that we cannot use all data for both because then we would not really measure the generalization ability but only how well the model can memorize the data.

Additionally, it is good practice to additionally use a validation dataset. This is data used to get feedback on the model during training or for comparing the performance of the same model but for different hyper-parameters, such as model size. The final test data is then only used for final checking and reporting of model performance.

## Capacity, underfitting and overfitting
* The capacity of a network refers to the range or scope of the types of functions that the model can approximate
* Too high capacity for a given problem can lead to overfitting
* Not enough capacity leads to underfitting


We have already seen the concepts of capacity, underfitting and overfitting. Let's look at these concepts in the context of neural networks. The capacity of a network refers to the range or scope of the types of functions that the model can approximate. Informally, a model's capacity is its ability to fit a wide variety of functions.
Too high capacity for a given problem can lead to overfitting. Not enough capacity leads to underfitting.

## Capacity, underfitting and overfitting


We need to make sure we choose the capacity that achieves the best generalization error. It is often the case that with more capacity we start overfitting.

## Regularization
With millions of parameters for large neural networks, regularization becomes crucial not to overfit

Common types of regularization include:
* Early stopping
* Dropout
* Weight decay
* Data augmentation

In particular, millions of parameters in large neural networks make it relatively easy to overfit. How do we deal with this if we still want to use large powerful models? We use regularization!

There are various ways to regularize neural networks. The most common ones include early stopping, dropout, weight decay and data augmentation.

## Early stopping
* Training neural networks for too long often leads to worse generalization
* Solution: use early stopping!


Training neural networks for too long often leads to worse generalization. The solution for this is simple: stop training when generalization becomes worse.

## Early stopping - details
How to do early stopping?
* Measure the generalization error on a validation set periodically during training
   * E.g. after each epoch
* Use the model that had the best validation error as the final model
* In practice you save the parameters of the best model so far and load the parameters of the best model in the end
* Could also do early stopping dynamically - e.g. when there was no improvement for given number of iterations

So how do we do early stopping? We measure the generalization error on a validation set periodically during training, e.g. after each epoch. We use the model that had the best validation error as the final model.
In practice we save the parameters of the best model so far and load the parameters of the best model in the end.
We could also do early stopping dynamically - e.g. when there was no improvement for given number of iterations.

## Dropout
Randomly remove a fraction of neurons each time a gradient it calculated
* Simulates training with many different networks
* Leads to more robust features



Dropout randomly removes a fraction of neurons each time a gradient it calculated. It simulates training with many different networks and leads to more robust features.

## Weight decay
* Penalize large parameter values, also known as $L_2$ regularization
* Instead of the original loss, minimize the following objective
$$L(w)+\lambda\sum_{j=1}^dw_j^2,$$
where $\lambda$ is a parameter and $d$ is the number of weights in the network

Weight decay - also known as L2 regularization - affects how large the weights can become. The key difference is to add the squared values of the weights to the loss function, scaled appropriately using parameter $\lambda$.

## Data augmentation

Finally, data augmentation simply adds new data points by doing various transformations of the data: for example by adding noise, rotating the image, flipping it or cropping it.
